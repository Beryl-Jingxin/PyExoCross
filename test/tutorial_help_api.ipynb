## 6. Result-specific help after calculating

The next cells use the MgH setup from `test_api_exomol.py`. They perform one small partition-function calculation so `result.help()` can show an actual result.

```python
COMMON = dict(
    database='ExoMol',
    molecule='MgH',
    isotopologue='24Mg-1H',
    dataset='XAB',
    species_id=501,
    read_path='/Users/beryl/Academic/Data/Database/ExoMol/',
    save_path='/Users/beryl/Academic/Data/pyexocross/',
    cache='parquet',
)

data = px.load(
    **COMMON,
    ncputrans=4, ncpufiles=1, chunk_size=500000, device='CPU',
    wn_wl='WN', wn_wl_unit='cm-1',
    min_range=0, max_range=30000, unc_filter=0.5,
    log='none', verbose=False,
)

result = px.partition_functions(
    data=data, ntemp=100, tmax=5000,
    output='memory', log='none', verbose=False,
)
```

```python
result.help()
result.summary()
```

## 7. Discover attributes and parameters

In Spyder or IPython, type `result.` and press Tab. Programmatically, filter `dir(result)` to list public attributes.

```python
public = [name for name in dir(result) if not name.startswith('_')]
print(public)

print(result.params)
print('ntemp:', result.params['ntemp'])
parameter_dict = result.params.to_dict()
print('parameter keys:', sorted(parameter_dict))
```

## 8. Selection help patterns

Use no selector for a single unconditional table, T/P for spectra, and `product=` for combined output.

```python
partition_table = result.select()
print(partition_table.head())

# Cross-section pattern:
# xsec = result.select(T=300, P=1.0)
# matches = xsec_result.select(T=300, squeeze=False)

# Combined patterns:
# stick = combined_result.select(T=300, product='stick_spectra')
# xsec = combined_result.select(T=300, P=1.0, product='cross_section')

# Two-temperature NLTE pattern:
# spectrum = nlte_result.select(Tvib=3000, Trot=296)
```

## 9. Common mistakes

- `result = px.cross_sections` stores the function. Use parentheses: `result = px.cross_sections(...)`.
- `output='files'` returns `None`; use `memory` or `both` to obtain Result.
- A combined result requires `product=`.
- Stick spectra do not use pressure.
- If several conditions match, add selectors or use `squeeze=False`.
- Conversion supports file output only.

## 10. Help regression checks
